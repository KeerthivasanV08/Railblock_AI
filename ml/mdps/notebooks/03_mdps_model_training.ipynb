# 03. MDPS Machine Learning Model Training & Evaluation

Trains the GradientBoostingRegressor model for Multi-Variable Criticality Matrix Priority Scoring (MDPS).

Features: `sev_num`, `overdue_days`, `traffic_num`, `deferred_count`  
Target: `actual_priority_rank`

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path('../../..').resolve()))

import pandas as pd
import numpy as np
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.model_selection import train_test_split

train_df = pd.read_csv(Path('../../../data/features/mdps_features/train.csv'))
val_df = pd.read_csv(Path('../../../data/features/mdps_features/validation.csv'))
test_df = pd.read_csv(Path('../../../data/features/mdps_features/test.csv'))

print(f'Train rows: {len(train_df):,}, Val rows: {len(val_df):,}, Test rows: {len(test_df):,}')
```

## 1. Fit Model & Evaluate

```python
features = ['sev_num', 'overdue_days', 'traffic_num', 'deferred_count']
target = 'actual_priority_rank'

scaler = StandardScaler()
X_train = scaler.fit_transform(train_df[features])
y_train = train_df[target]

X_test = scaler.transform(test_df[features])
y_test = test_df[target]

model = GradientBoostingRegressor(n_estimators=150, learning_rate=0.08, max_depth=4, random_state=42)
model.fit(X_train, y_train)

preds = model.predict(X_test)
print(f'MAE:  {mean_absolute_error(y_test, preds):.4f}')
print(f'RMSE: {np.sqrt(mean_squared_error(y_test, preds)):.4f}')
print(f'R2:   {r2_score(y_test, preds):.4f}')
```
