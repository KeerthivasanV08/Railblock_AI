# 05. Constraint Engine & OR-Tools MILP Optimization

Demonstrates the operational constraint filtering (traffic, machine, crew, duration, spatial) and mathematical optimization via Google OR-Tools (SCIP / CBC solver).

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path('../../../backend').resolve()))

import pandas as pd
from app.services.optimization.constraints import ConstraintEngine
from app.services.optimization.milp_solver import OptimizationEngine

constraint_engine = ConstraintEngine()
opt_engine = OptimizationEngine()
```

```python
tasks = pd.read_csv(Path('../../../data/processed/clustered_tasks.csv'))
feas_df = constraint_engine.check_feasibility(tasks)
print(f'Feasible: {feas_df["overall_feasible"].sum():,}, Infeasible: {(~feas_df["overall_feasible"]).sum():,}')

selected_blocks, metrics = opt_engine.optimize_blocks(feas_df)
print(f'Optimizer Status: {metrics.get("status")}, Selected Blocks: {len(selected_blocks)}')
```
