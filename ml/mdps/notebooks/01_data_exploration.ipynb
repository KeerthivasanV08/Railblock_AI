# 01. RailBlock AI — Raw & Unified Dataset Exploration

This notebook inspects the raw datasets (TMS, SMMS, TDMS defects, network geometry, train timetable, live delays, and resources) and analyzes data quality, distributions, and missing values.

```python
import pandas as pd
from pathlib import Path

DATA_DIR = Path('../../../data')
print('Data directory:', DATA_DIR.resolve())
```

## 1. Defect Dataset Inventory

```python
tms = pd.read_csv(DATA_DIR / 'raw/defects/tms_defects.csv')
smms = pd.read_csv(DATA_DIR / 'raw/defects/smms_defects.csv')
tdms = pd.read_csv(DATA_DIR / 'raw/defects/tdms_defects.csv')

print(f'TMS Defects:  {len(tms):,} rows, cols={tms.columns.tolist()}')
print(f'SMMS Defects: {len(smms):,} rows, cols={smms.columns.tolist()}')
print(f'TDMS Defects: {len(tdms):,} rows, cols={tdms.columns.tolist()}')
```

## 2. Severity & Department Distribution

```python
print('TMS Severity Class:\n', tms['severity_class'].value_counts())
print('\nSMMS Severity Class:\n', smms['severity_class'].value_counts())
print('\nTDMS Severity Class:\n', tdms['severity_class'].value_counts())
```

## 3. Network Geometry & Traffic Feeds

```python
sections = pd.read_csv(DATA_DIR / 'raw/network/block_sections.csv')
timetable = pd.read_csv(DATA_DIR / 'raw/traffic/train_timetable.csv')
delays = pd.read_csv(DATA_DIR / 'raw/traffic/live_train_delays.csv')

print(f'Block Sections: {len(sections)} sections')
print(f'Timetable Trains: {len(timetable)} schedules')
print(f'Live Delays: {len(delays)} records (contains duplicate checks)')
```
