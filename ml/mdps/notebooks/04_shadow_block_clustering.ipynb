# 04. Deterministic Shadow Block Clustering

Groups spatially proximal maintenance tasks (within 2.0 km chainage threshold) into consolidated shadow blocks to minimize train traffic disruptions.

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path('../../../backend').resolve()))

import pandas as pd
from app.services.clustering.shadow_block_service import ShadowBlockService

clustering_service = ShadowBlockService()
tasks_df = pd.read_csv(Path('../../../data/processed/scored_tasks.csv'))
print(f'Loaded {len(tasks_df)} scored tasks.')
```

```python
clustered_df = clustering_service.cluster_tasks(tasks_df)
print(f'Total Clusters Generated: {clustered_df["cluster_id"].nunique()}')
print(f'Integrated Block Candidates: {clustered_df["integrated_block_candidate"].sum()}')
```
